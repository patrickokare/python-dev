💪 Challenge 1: Gym Tracker (Lists, OOP, If Statements)

Track a user’s workouts.

Class: GymMember
	•	__init__(name) → store name and empty list of workouts
	•	add_workout(workout) → only add if it’s not already in list
	•	is_active() → return True if more than 3 workouts

⸻

```python
class GymMember:
  # 1.Constructor method - runs when an object is created
  def __init__(self, name):
    self.name = name
    self.workouts = []

# 2.method 
  def add_workout(self, workout):
      if workout not in self.workouts:
        self.workouts.append(workout)
        
        print(f"Adding {workout} for {self.name}")

        
# 3.method 
  def is_active(self):
    if len(self.workouts) > 3:
      return True
    else:
      return False


  def summary(self):
    print(f"{self.name} has done {len(self.workouts)} workouts: {self.workouts}")
  


john = GymMember("John") #We are creating an instance of the Class
john.add_workout("Pushups") #adding each of our 3 exercises for John
john.add_workout("Squats")
john.add_workout("Jogging")

print(john.workouts)
print(john.is_active())
john.summary()

```
